```python
import numpy as np

def get_list_random(a: int, b: int, length: int):
    """
    Функция возвращает список равномерно распределенных чисел между значениями а и б.
    :param a: наименьшее число
    :param b: наибольшее число
    :param length: количество значений в списке
    :return: список равномерно распределенных значений
    """
    if a > b:
        raise ValueError('Параметр a, должен быть меньше, чем параметр b')

    if length <= 0:
        return ValueError('Параметр length должен быть больше 0')

    return np.random.uniform(a, b, size=length)

print(get_list_random(1, 10, 30))
```

```python
def find_duplicates(args):
    """
    Функция считает количество уникальных значений (строит частотный ряд)
    :param args: список, для обработки
    :return: справочник в формате {значение: количество вхождений}
    """
    if not isinstance(args, list):
        raise ValueError('Переданный параметр должен являться списком')

    result = {}

    for item in args:
        if item not in result:
            result[item] = 1
        else:
            result[item] += 1

    return result

print(find_duplicates([1, 2, 3, 1, 2, 2]))
```

```python
def get_statistic(args):
    """
    Функция определяет статистическую информацию по переданному списку и сохраняет ее в справочник
    :param args: список
    :return: справочник с ключами min, max, mean, median, mode
    """
    if not isinstance(args, list):
        raise ValueError('Переданный параметр должен являться списком')

    array = np.array(args)

    result = {"min": array.min(), "max": array.max(), "mean": array.mean(), "median": np.median(array)}

    values, counts = np.unique(array, return_counts=True)
    idx = np.argmax(counts)
    result["mode"] = values[idx]

    return result

print(get_statistic([1, 2, 3, 1, 2, 2]))

# с ошибкой
# print(get_statistic(2))
```
